# HQFSF: Data Exploration

## Hybrid Quantum Feature Selection Framework (HQFSF)

This notebook performs exploratory data analysis (EDA) on the Breast Cancer Wisconsin dataset used in the HQFSF framework.

**Objectives**

- Load the dataset
- Explore its structure
- Analyze feature statistics
- Detect missing values
- Check duplicate records
- Visualize class distribution
- Study feature correlations

```python
import warnings

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from sklearn.datasets import load_breast_cancer

warnings.filterwarnings("ignore")
```

```python
dataset = load_breast_cancer()

df = pd.DataFrame(
    dataset.data,
    columns=dataset.feature_names,
)

df["target"] = dataset.target
```

```python
print("Dataset Shape")
print(df.shape)

print("\nFirst Five Rows")
display(df.head())

print("\nColumns")
display(df.columns)
```

```python
df.info()
```

```python
df.describe().T
```

```python
missing = df.isnull().sum()

display(
    missing[missing > 0]
)
```

```python
duplicates = df.duplicated().sum()

print(f"Duplicate Rows : {duplicates}")
```

```python
counts = df["target"].value_counts()

plt.figure(figsize=(6,4))

plt.bar(
    ["Malignant", "Benign"],
    counts.sort_index()
)

plt.title("Class Distribution")

plt.xlabel("Class")

plt.ylabel("Samples")

plt.tight_layout()

plt.show()
```

```python
corr = df.corr(numeric_only=True)

plt.figure(figsize=(12,10))

plt.imshow(
    corr,
    aspect="auto",
)

plt.colorbar()

plt.title("Correlation Matrix")

plt.tight_layout()

plt.show()
```

```python
variance = df.var()

variance.sort_values(
    ascending=False
).head(10)
```

```python
df.hist(
    figsize=(18,15)
)

plt.tight_layout()

plt.show()
```

```python
print("Benign Samples")

print((df["target"] == 1).sum())

print()

print("Malignant Samples")

print((df["target"] == 0).sum())
```

```python
summary = pd.DataFrame({
    "Property": [
        "Samples",
        "Features",
        "Missing Values",
        "Duplicate Rows",
        "Classes"
    ],
    "Value": [
        len(df),
        len(dataset.feature_names),
        df.isnull().sum().sum(),
        df.duplicated().sum(),
        df["target"].nunique()
    ]
})

display(summary)
```
