# HQFSF - Data Preprocessing

**Hybrid Quantum Feature Selection Framework Using Variational Quantum Circuits**

This notebook performs the preprocessing steps required before quantum feature selection and classical machine learning.

### Objectives

- Load dataset
- Inspect data
- Handle missing values
- Remove duplicates
- Encode categorical variables
- Scale numerical features
- Split dataset
- Save processed data

```python
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import (
    LabelEncoder,
    StandardScaler
)

import matplotlib.pyplot as plt
```

```python
DATASET_PATH = "../datasets/raw/dataset.csv"

df = pd.read_csv(DATASET_PATH)

print(df.shape)

df.head()
```

```python
df.info()
```

```python
df.describe(include="all")
```

```python
df.isnull().sum()
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())
```

```python
categorical_columns = df.select_dtypes(exclude=np.number).columns

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])
```

```python
print("Before:", len(df))

df = df.drop_duplicates()

print("After :", len(df))
```

```python
encoder = LabelEncoder()

for column in categorical_columns:

    df[column] = encoder.fit_transform(df[column])
```

```python
X = df.iloc[:, :-1]

y = df.iloc[:, -1]

print(X.shape)

print(y.shape)
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns
)

X_scaled.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y,
)

print(X_train.shape)

print(X_test.shape)
```

```python
processed = X_scaled.copy()

processed["target"] = y.values

processed.to_csv(
    "../datasets/processed/processed_dataset.csv",
    index=False,
)

print("Processed dataset saved successfully.")
```

```python
train = X_train.copy()

train["target"] = y_train.values

test = X_test.copy()

test["target"] = y_test.values

train.to_csv(
    "../datasets/processed/train.csv",
    index=False,
)

test.to_csv(
    "../datasets/processed/test.csv",
    index=False,
)

print("Train and Test datasets saved.")
```
