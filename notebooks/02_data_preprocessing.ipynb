# HQFSF: Data Preprocessing

## Hybrid Quantum Feature Selection Framework (HQFSF)

This notebook performs data preprocessing before quantum feature selection.

### Objectives

- Load the dataset
- Validate the data
- Remove duplicates
- Handle missing values
- Encode target labels
- Scale numerical features
- Split training and testing data
- Save the processed dataset

```python
import warnings

import joblib
import pandas as pd

from pathlib import Path

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler

warnings.filterwarnings("ignore")
```

```python
dataset = load_breast_cancer()

df = pd.DataFrame(
    dataset.data,
    columns=dataset.feature_names,
)

df["target"] = dataset.target

print(df.head())
```

```python
print("Shape :", df.shape)

print()

print(df.info())
```

```python
missing = df.isnull().sum()

display(
    missing[missing > 0]
)
```

```python
duplicates = df.duplicated().sum()

print("Duplicate Rows :", duplicates)
```

```python
df = df.drop_duplicates()

print("Updated Shape :", df.shape)
```

```python
X = df.drop(
    columns=["target"]
)

y = df["target"]

print(X.shape)

print(y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y,
)

print("Training Samples :", len(X_train))

print("Testing Samples :", len(X_test))
```

```python
scaler = MinMaxScaler()
```

```python
X_train_scaled = scaler.fit_transform(
    X_train
)
```

```python
X_test_scaled = scaler.transform(
    X_test
)
```

```python
X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns=X.columns,
)

X_test_scaled = pd.DataFrame(
    X_test_scaled,
    columns=X.columns,
)
```

```python
Path("saved_models").mkdir(
    exist_ok=True
)

joblib.dump(
    scaler,
    "saved_models/minmax_scaler.joblib",
)
```

```python
Path("datasets/processed").mkdir(
    parents=True,
    exist_ok=True,
)

X_train_scaled.to_csv(
    "datasets/processed/X_train.csv",
    index=False,
)

X_test_scaled.to_csv(
    "datasets/processed/X_test.csv",
    index=False,
)

y_train.to_csv(
    "datasets/processed/y_train.csv",
    index=False,
)

y_test.to_csv(
    "datasets/processed/y_test.csv",
    index=False,
)
```

```python
summary = pd.DataFrame(
    {
        "Property": [
            "Original Samples",
            "Training Samples",
            "Testing Samples",
            "Features",
        ],
        "Value": [
            len(df),
            len(X_train_scaled),
            len(X_test_scaled),
            X_train_scaled.shape[1],
        ],
    }
)

display(summary)
```

```python
print("Minimum Value")

print(X_train_scaled.min().min())

print()

print("Maximum Value")

print(X_train_scaled.max().max())
```
